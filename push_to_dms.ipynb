```python
from database.dependencies import DB_DEPENDENCY
from sqlalchemy.orm import Query
from sqlalchemy import text


def query_sql(query:str):
    from database.session import SessionLocal
    # To list tables in PostgreSQL, use the information_schema.tables view:
    with SessionLocal() as db:
        result = db.execute(text(query))
        # UPDATE/INSERT/DELETE don't return rows - use rowcount instead
        if result.returns_rows:
            return result.fetchall()
        db.commit()
        return result.rowcount
```

```python
# ORM (SQLAlchemy) model for public.disposition_qa (generated from CREATE TABLE DDL)

from sqlalchemy import (
    Column, String, Integer, BigInteger, Float, Text, ForeignKey, Index,
    DateTime, JSON, Boolean, text
)
from sqlalchemy.dialects.postgresql import JSONB
from sqlalchemy.orm import declarative_base
import os

Base = declarative_base()

class DispositionQA(Base):
    __tablename__ = 'disposition_qa'
    __table_args__ = (
        Index('idx_dqa_alloted_to_uid', 'alloted_to', postgresql_using='btree'),
        Index('idx_dqa_client_name_uid', 'client_name', postgresql_using='btree'),
        Index('idx_dqa_current_disposition_uid', 'current_disposition', postgresql_using='btree'),
        Index('idx_dqa_date_uid', text('date DESC'), postgresql_using='btree'),
        Index('idx_dqa_transcript_gin_uid', 'transcript', postgresql_using='gin'),
    )

    id = Column(BigInteger, primary_key=True, autoincrement=True)
    uid = Column(String(100), nullable=False)
    transcript = Column(JSONB, nullable=True)
    current_disposition = Column(String(255), nullable=True)
    ts = Column(JSONB, nullable=True)
    alloted_to = Column(Integer, nullable=True)
    updated_disposition = Column(String(255), nullable=True)
    language = Column(String(255), nullable=True)
    file_name = Column(String(500), nullable=True)
    reason = Column(Text, nullable=True)
    date = Column(DateTime(timezone=True), server_default=text("(now() AT TIME ZONE 'Asia/Kolkata')"), nullable=True)
    client_name = Column(String(255), nullable=True)
    llm = Column(JSONB, nullable=True)
    duration = Column(Float, nullable=True)
    viewed = Column(Boolean, nullable=False, server_default=text('false'))
    viewed_at = Column(DateTime(timezone=True), nullable=True)
```

```python
# Prepare data for insert
from sqlalchemy.orm.session import Session


_data_df = df_merged[['uid', 'chat', 'Disposition_code', 'recording', 'file_name', 'date', 'client_name', 'duration']]

# Database connection
from sqlalchemy.orm import sessionmaker
from sqlalchemy import create_engine

db_host = os.getenv('DB_HOST')
db_port = os.getenv('DB_PORT')
db_name = os.getenv('DB_NAME')
db_user = os.getenv('DB_USER')
db_password = os.getenv('DB_PASSWORD')

DATABASE_URL = f"postgresql+psycopg2://{db_user}:{db_password}@{db_host}:{db_port}/{db_name}"
engine = create_engine(DATABASE_URL)
SessionLocal = sessionmaker[Session](autocommit=False, autoflush=False, bind=engine)
session = SessionLocal()
```

```python
# Always INSERT (allow repeated uid). Debug row-level failures.
import json
import ast
import pandas as pd

lang = 'hindi'
def to_json(val):
    """Convert Python object/string to valid JSON for JSONB column."""
    if val is None or (isinstance(val, float) and pd.isna(val)):
        return None
    if isinstance(val, str):
        # Already JSON?
        try:
            json.loads(val)
            return val
        except Exception:
            pass
        # Python repr string like "[{'role': ...}]"?
        try:
            parsed = ast.literal_eval(val)
            return json.dumps(parsed, ensure_ascii=False)
        except Exception:
            pass
    return json.dumps(val, ensure_ascii=False)


def to_float(val):
    if val is None or (isinstance(val, float) and pd.isna(val)):
        return None
    try:
        return float(val)
    except Exception:
        return None


inserted = 0

for i, row in enumerate(_data_df.itertuples(index=False), start=1):
    # row: uid, chat, Disposition_code, file_name, date, client_name, duration
    uid_val = getattr(row, 'uid')
    disp_val = getattr(row, 'Disposition_code')
    dur_raw = getattr(row, 'duration')
    dur_val = to_float(dur_raw)

    try:
        new_row = DispositionQA(
            uid=uid_val,
            language=lang,
            transcript=to_json(getattr(row, 'chat')),
            current_disposition=disp_val,
            file_name=getattr(row, 'file_name'),
            date=getattr(row, 'date'),
            client_name=getattr(row, 'client_name'),
            duration=dur_val,
        )
        session.add(new_row)
        session.flush()  # force DB insert now
        inserted += 1

        if i % 1000 == 0:
            session.commit()
            print(f"Committed {i} rows...")

    except Exception as e:
        session.rollback()
        print("FAILED at row", i)
        print("uid:", uid_val)
        print("Disposition_code:", disp_val)
        print("duration raw:", dur_raw, "type:", type(dur_raw))
        print("duration coerced:", dur_val, "type:", type(dur_val))
        raise

session.commit()
session.close()
print(f"Done! Inserted: {inserted}") 
```

```python
import asyncio
import aiohttp

# Progress bar selection: notebook tqdm can fail at runtime
# (e.g., when ipywidgets' IProgress isn't available).
try:
    from tqdm.notebook import tqdm as tqdm_notebook
    tqdm_bar = tqdm_notebook
except Exception:
    tqdm_bar = None

if tqdm_bar is None:
    try:
        from tqdm import tqdm as tqdm_terminal
        tqdm_bar = tqdm_terminal
    except ImportError:
        # Fallback if tqdm is not installed at all
        def tqdm_bar(x, **kwargs):
            return x

async def async_check_recording(session, url, idx, df):
    """
    Asynchronously checks if a single audio recording exists.
    """
    try:
        async with session.head(url, timeout=aiohttp.ClientTimeout(total=5)) as response:
            if response.status == 200:
                df.at[idx, 'exit'] = True
                print(f"{idx}: working (audio exists)")
            else:
                df.at[idx, 'exit'] = False
                print(f"{idx}: not working (no audio)")
    except Exception as e:
        df.at[idx, 'exit'] = False
        print(f"{idx}: not working (error: {e})")

async def check_recordings_async(df, batch_size=100):
    """
    Asynchronously checks if the audio recordings exist at the given URLs.
    Processes in batches to keep memory/network usage controlled.
    """
    connector = aiohttp.TCPConnector(limit=20)  # You can tune this
    async with aiohttp.ClientSession(connector=connector) as session:
        indices_urls = list(df[['f_recording']].itertuples(name=None))
        total = len(indices_urls)
        batch_iterable = range(0, total, batch_size)
        try:
            batch_range = tqdm_bar(batch_iterable, desc='Checking recordings')
        except ImportError:
            # tqdm.notebook can raise ImportError when ipywidgets' IProgress isn't available.
            from tqdm import tqdm as tqdm_terminal
            batch_range = tqdm_terminal(batch_iterable, desc='Checking recordings')

        for batch_start in batch_range:
            tasks = []
            batch = indices_urls[batch_start:batch_start+batch_size]
            for idx, (row_idx, url) in enumerate(batch, batch_start):
                tasks.append(async_check_recording(session, url, row_idx, df))
            await asyncio.gather(*tasks)

# --- Handle Jupyter/Notebook async event loop properly ---
import sys

def run_check_recordings_async(df):
    try:
        loop = asyncio.get_running_loop()
    except RuntimeError:  # No event loop running
        loop = None

    if loop and loop.is_running():
        # We are probably in a notebook cell, and event loop already running.
        # Try to workaround ImportError from tqdm.notebook/ipw using plain tqdm.
        # Use nest_asyncio to allow nested loop for notebooks
        try:
            import nest_asyncio
            nest_asyncio.apply()
        except ImportError:
            pass  # nest_asyncio is only needed in some notebook cases

        # Use ensure_future and run_until_complete for notebook compatibility
        task = asyncio.ensure_future(check_recordings_async(df))
        try:
            loop.run_until_complete(task)
        except ImportError as e:
            print(f"tqdm notebook progress bar failed due to missing ipywidgets / IProgress. "
                  f"Rerun code after installing ipywidgets or fall back to terminal tqdm!")
            return
    else:
        # Safe to use asyncio.run as no loop running
        asyncio.run(check_recordings_async(df))

# Run the function to update the DataFrame asynchronously
run_check_recordings_async(df)
```
