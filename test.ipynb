```python
from database.dependencies import DB_DEPENDENCY
from sqlalchemy.orm import Query
from sqlalchemy import text


def query_sql(query:str):
    from database.session import SessionLocal
    # To list tables in PostgreSQL, use the information_schema.tables view:
    with SessionLocal() as db:
        result = db.execute(text(query))
        # UPDATE/INSERT/DELETE don't return rows - use rowcount instead
        if result.returns_rows:
            return result.fetchall()
        db.commit()
        return result.rowcount
```

```python
from database.dependencies import DB_DEPENDENCY

from sqlalchemy.ext.declarative import declarative_base
from sqlalchemy import (
    Column, Integer, BigInteger, String, Text, Float, Boolean, JSON, DateTime, Numeric, func, text
)
import sqlalchemy.dialects.postgresql as pg

Base = declarative_base()

class SpandanaSphoortyData(Base):
    __tablename__ = 'spandana_sphoorty_data'
    __table_args__ = {"schema": "public"}

    s_no = Column(Integer, primary_key=True, autoincrement=True)
    json_data = Column(pg.JSONB)
    disposition = Column(Text)
    created_at = Column(DateTime(timezone=True), server_default=text("(CURRENT_TIMESTAMP AT TIME ZONE 'Asia/Kolkata')"))
    language = Column(Text)
    is_disposition_available = Column(Boolean, server_default=text('false'))
    transcript = Column(Text)
    is_transcript_available = Column(Boolean, server_default=text('false'))
    contact_to = Column(Text)
    contact_from = Column(Text)
    direction = Column(Text)
    call_start_ts = Column(Float)
    call_end_ts = Column(Float)
    chat = Column(pg.JSONB)
    provider = Column(Text)
    call_id = Column(Text)
    agent_id = Column(Text)
    duration = Column(Float)
    meta_data = Column(pg.JSONB)
    recording = Column(Text)
    model = Column(Text)
    cost = Column(Float)
    is_duplicate = Column(Boolean, server_default=text('false'))
    campaign_id = Column(Integer)
    session_id = Column(String)
    call_status = Column(String)
    error_message = Column(String)
    uid = Column(String)

class SpandanaDebtCollectionReport(Base):
    __tablename__ = 'spandana_debt_collection_report'
    __table_args__ = {"schema": "public"}

    id = Column(BigInteger, primary_key=True, autoincrement=True)
    import_batch_id = Column(pg.UUID, server_default=text('gen_random_uuid()'), nullable=False)
    source_file_name = Column(String(255), nullable=False)
    imported_at = Column(DateTime(timezone=True), server_default=text("(now() AT TIME ZONE 'Asia/Kolkata')"), nullable=False)
    borrower_code = Column(String(100), nullable=False)
    vendor_name = Column(String(255))
    connected_status = Column(String(100))
    recording_link = Column(Text)
    duration_seconds = Column(Integer)
    disposition = Column(String(255))
    language = Column(String(100))
    zone = Column(String(100))
    svp_state = Column(String(100))
    region = Column(String(100))
    area = Column(String(100))
    cluster = Column(String(100))
    branch_code = Column(String(50))
    branch_name = Column(String(255))
    repayment_frequency = Column(String(100))
    center_id = Column(String(100))
    center_name = Column(String(255))
    group_name = Column(String(255))
    village_name = Column(String(255))
    borrower_name = Column(String(255))
    spouse_name = Column(String(255))
    phone_number = Column(String(50))
    no_of_loans = Column(Integer)
    pos = Column(Numeric(18, 2))
    od = Column(Numeric(18, 2))
    emi = Column(Numeric(18, 2))
    dpd = Column(Integer)
    last_repayment_paid_date = Column(String(50))
    npa_date_as_fimo = Column(String(50))
    old_new = Column(String(50))
    alternate_mobile_number = Column(String(50))
    phase = Column(String(100))
    created_at = Column(DateTime(timezone=True), server_default=text("(now() AT TIME ZONE 'Asia/Kolkata')"), nullable=False)

```

```python
import numbers

def ensure_py_types(val):
    """
    Converts numpy numeric types to built-in Python types.
    Returns None for NaN values.
    """
    if pd.isna(val):
        return None
    # Convert numpy.integer/float to built-in types
    if isinstance(val, numbers.Integral):
        return int(val)
    if isinstance(val, numbers.Real):
        return float(val)
    return val

def get_float(val):
    """Attempts to coerce to float, returns None if not possible (empty, strange text, etc.)."""
    try:
        # Will be nan if not convertible, caught by ensure_py_types
        return ensure_py_types(pd.to_numeric(val, errors='coerce'))
    except Exception:
        return None

def get_int(val):
    """Attempts to coerce to int, returns None if not possible (empty, strange text, etc.)."""
    try:
        result = pd.to_numeric(val, errors='coerce')
        return int(result) if not pd.isna(result) else None
    except Exception:
        return None

def safe_str(val):
    if pd.isna(val):
        return None
    value = str(val).strip()
    if value.lower() in ('nan', 'none', ''):
        return None
    return value

# For all strictly numeric fields, ensure only proper numbers (no accidental text!); 
# for all text fields, always treat as string or None.
with SessionLocal() as session:
    records = []
    for _, row in df.iterrows():
        try:
            # Numeric fields - ensure only int or float or None
            duration_seconds = get_float(row['Duration (in Seconds)'])
            center_id = get_int(row['CenterID'])
            no_of_loans = get_int(row['NOof_Loans'])
            pos = get_float(row['POS'])
            od = get_float(row['OD '])
            emi = get_float(row['EMI'])
            dpd = get_int(row['DPD'])

            record = SpandanaDebtCollectionReport(
                source_file_name='SSFL__Debt Collection Report_27th Jan 2026.xlsx',
                borrower_code=safe_str(row['BorrowerCode']),
                vendor_name='FGPL',
                connected_status=safe_str(row['Connected Status']),
                recording_link=safe_str(row['Recording Link']),
                duration_seconds=duration_seconds,
                disposition=safe_str(row['Disposition (Best Accuracy)']),
                language=safe_str(row['Language']),
                zone=safe_str(row['Zone']),
                svp_state=safe_str(row['SVP State']),
                region=safe_str(row['Region']),
                area=safe_str(row['Area']),
                cluster=safe_str(row['Cluster']),
                branch_code=safe_str(row['Branch Code']),
                branch_name=safe_str(row['BranchName']),
                repayment_frequency=safe_str(row['Repayment Frequency']),
                center_id=center_id,
                center_name=safe_str(row['CenterName']),
                group_name=safe_str(row['GroupName']),
                village_name=safe_str(row['Village Name']),
                borrower_name=safe_str(row['BorrowerName']),
                spouse_name=safe_str(row['Spouse Name']),
                phone_number=safe_str(row['Phone Numeber']),
                no_of_loans=no_of_loans,
                pos=pos,
                od=od,
                emi=emi,
                dpd=dpd,
                last_repayment_paid_date=safe_str(row['Last Repayment Paid Date']),
                npa_date_as_fimo=safe_str(row['NPA_Date_as_fimo']),
                old_new=safe_str(row['OLD /New']),
                alternate_mobile_number=safe_str(row['AlterNateMobileNumber']),
                phase=safe_str(row['Phase'])
            )
            records.append(record)
        except Exception as e:
            # Log bad row and skip
            print(f"Skipping row {_} due to error: {e}")
            continue
    session.add_all(records)
    session.commit()
```

```python
import json

# Custom JSON loader for possibly malformed lines
def safe_json_load(line):
    try:
        return json.loads(line)
    except Exception as e:
        print(f"Could not parse line (starts with {line[:30]}...): {e}")
        return None

data = []
filename = "/Users/apple/Downloads/campaign_data-4.json"
with open(filename, "r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        item = safe_json_load(line)
        if item is None:
            continue
        # If 'meta_data' is a string, decode it when possible
        if 'meta_data' in item and isinstance(item['meta_data'], str):
            try:
                item['meta_data'] = json.loads(item['meta_data'])
            except Exception:
                # If nested 'meta_data' string is not valid JSON, just leave it
                pass
        data.append(item)

# Print a few examples of meta_data (to check unicode decoding)
for i, item in enumerate(data[:5]):
    print(item.get('meta_data'))

# Example of accessing the 'Name' key
for item in data[:5]:
    meta = item.get('meta_data', {})
    name = meta.get("Name") if isinstance(meta, dict) else None
    print("Name:", name)
```
